# --- WEEK 13: Cleaning & Analytics ---

```python

# 1. Inspect reservation status, cancellation flags, lead time
print(df['is_canceled'].value_counts())
print(df['reservation_status'].value_counts())
print(df['lead_time'].describe())
```

```python
# 2. Handle missing data
# Check missing %
print(df.isnull().sum() / len(df) * 100)
```

```python
# 11. Missing count 
df['reservation_status'].isnull().sum()
```

```python
df.duplicated().sum() 
df[df.duplicated()] 
```

```python
# 3. Feature Engineering
df['total_stay'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']
df['total_guests'] = df['adults'] + df['children'] + df['babies']
```

```python
# 4. Cancellation % across hotel types
cancel_rate = df.groupby('hotel')['is_canceled'].mean()*100
print(cancel_rate)
```

```python
# 5. Detect outliers - Lead time and ADR
plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
sns.boxplot(x=df['lead_time'])
plt.title('Lead Time Outliers')
plt.figure(figsize=(6,4))
sns.barplot(data=df, x='hotel', y='is_canceled')
plt.title('Cancellation %: City Hotel vs Resort Hotel')
plt.show()
```

```python
# --- WEEK 14: Visualization & Insights ---
```

```python
sns.set_style("whitegrid")
```

```python
# 1. Cancellation rate across market segments and hotel types
plt.figure(figsize=(8,5))
sns.barplot(data=df, x='market_segment', y='is_canceled', hue='hotel')
plt.title('Cancellation Rate by Market Segment & Hotel Type')
plt.xticks(rotation=30)
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.barplot(data=df, x='hotel', y='is_canceled')
plt.title('Cancellation %: City Hotel vs Resort Hotel')
plt.show()
```

```python
# 2. Lead time vs Cancellation
plt.figure(figsize=(6,4))
sns.boxplot(data=df, x='is_canceled', y='lead_time', hue='hotel')
plt.title('Lead Time: Canceled vs Non-Canceled')
plt.show()
```

```python
# 3. Monthly seasonality - Arrival trends and ADR
monthly_arrivals = df.groupby('arrival_date_month')['is_canceled'].count()
# Order months correctly
month_order = ['January','February','March','April','May','June','July','August','September','October','November','December']
monthly_arrivals = monthly_arrivals.reindex(month_order)
monthly_adr = df.groupby('arrival_date_month')['adr'].mean().reindex(month_order)
```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
monthly_arrivals.plot(marker='o')
plt.title('Monthly Arrival Trends')
plt.xticks(rotation=45)
```

```python
plt.subplot(1,2,2)
monthly_adr.plot(marker='o', color='orange')
plt.title('Monthly ADR Fluctuation')
plt.xticks(rotation=45)
plt.show()
```

```python
# 4. Final Insights Plots
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='arrival_date_month', order=month_order, hue='hotel')
plt.title('Seasonal Demand Surge')
plt.xticks(rotation=45)
plt.show()
```
