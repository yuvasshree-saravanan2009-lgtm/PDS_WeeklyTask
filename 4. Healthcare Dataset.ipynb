```python
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"])

df["Length of stay"] = (df["Discharge Date"] - df["Date of Admission"]).dt.days

df[["Age", "Billing Amount", "Room Number", "Length of stay"]].corr()
```

```python
df["Length of Stay"] = (
    pd.to_datetime(df["Discharge Date"]) -
    pd.to_datetime(df["Date of Admission"])
).dt.days
df[["Date of Admission","Discharge Date","Length of Stay"]].head()
```

```python
df["Admission Month"] = pd.to_datetime(
    df["Date of Admission"]
).dt.year
df[["Date of Admission","Admission Month"]].head()
```

```python
df.groupby("Admission Year")["Billing Amount"].mean()
```

```python
monthly_admissions=(
    df.groupby("Admission Month")["Name"]
    .count()
    .reset_index()
)
monthly_admissions
```

```python
plt.figure(figsize=(9,5))
sns.lineplot(
    data=monthly_admissions,
    x="Admission Month",
    y="Name",
    marker="o"
)
plt.title("Monthly Number of Admissions")
plt.xlabel("Admission North")
plt.ylabel("Number of Admissions")

plt.show()
```
